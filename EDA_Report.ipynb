# Exploratory Data Analysis (EDA) Report
Basic EDA steps using pandas and matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data.csv')
df.head()
```

```python
# Dataset info
df.info()
```

```python
# Summary statistics
df.describe()
```

```python
# Missing values
df.isnull().sum()
```

```python
# Correlation
df.corr(numeric_only=True)
```

```python
# Example plot
df.hist()
plt.show()
```

```python
# Save summary
df.describe().to_csv('eda_summary.csv')
```
