# Data Cleaning Notebook
Basic steps for cleaning a dataset using pandas.

```python
import pandas as pd
df = pd.read_csv('data.csv')
df.head()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Fill missing values
df = df.fillna(method='ffill')
```

```python
# Save cleaned data
df.to_csv('cleaned_data.csv', index=False)
```
